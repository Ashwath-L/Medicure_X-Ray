# 01. Exploratory Data Analysis (Chest X-Ray & CT Scans)

Analyze class distributions, image dimensions, aspect ratios, and intensity distributions.

```python
import os
from pathlib import Path
import cv2
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.config import XRAY_RAW_DIR

print("Raw X-Ray Directory:", XRAY_RAW_DIR)
```

```python
# Inspect class balance
counts = {}
for split in ["train", "test"]:
    for cls in ["NORMAL", "PNEUMONIA"]:
        p = XRAY_RAW_DIR / split / cls
        if p.exists():
            counts[f"{split}_{cls}"] = len(list(p.glob("*.*")))

pd.Series(counts).plot(kind="bar", color="teal", title="Class Distribution")
plt.ylabel("Count")
plt.show()
```
